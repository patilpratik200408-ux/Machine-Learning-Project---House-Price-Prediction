```python
df = pd.read_excel("../../Dataset/pune_house_price_10000_realistic.xlsx")
```

```python
encode = pd.get_dummies(df,columns=['location','property_type','furnishing'],dtype="int32")
df = pd.concat([df,encode],axis =1)
```

```python
x=encode[['area_sqft', 'bedrooms', 'bathrooms', 'age_years', 'floor',
       'total_floors', 'parking', 'location_Aundh',
       'location_Baner', 'location_Chinchwad', 'location_Hadapsar',
       'location_Hinjewadi', 'location_Kharadi', 'location_Kondhwa',
       'location_Kothrud', 'location_Pimpri', 'location_Viman Nagar',
       'location_Wagholi', 'location_Wakad', 'property_type_Apartment',
       'property_type_Independent House', 'property_type_Row House',
       'property_type_Villa', 'furnishing_Fully Furnished',
       'furnishing_Semi-Furnished', 'furnishing_Unfurnished']]
y = encode["price_lakhs"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=48)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
scaler.fit(x_train)
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```
